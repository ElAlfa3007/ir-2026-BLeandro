## Embeddings

```python
import matplotlib.pyplot as plt
```

```python
import gensim.downloader as api
model = api.load("word2vec-google-news-300")
```

```python
model
```

```python
vec = lambda word: model[word]
```

```python
x = "king"
```

```python
model.most_similar(x)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
```

```python
palabras = ['madrid', 'spain', 'paris', 'france']
v300 = [vec(word) for word in palabras]
x = vec('spain')-vec('madrid')-vec('paris')
v300.append(x)
v = pca.fit_transform(v300)
```

```python
plt.figure(figsize=(10,7))
plt.scatter(v[:,0], v[:,1])

plt.annotate('madrid', (v[0,0], v[0,1]))
plt.annotate('spain', (v[1,0], v[1,1]))
plt.annotate('paris', (v[2,0], v[2,1]))
plt.annotate('france', (v[3,0], v[3,1]))
plt.annotate('x', (v[4,0], v[4,1]))

plt.show()
```
